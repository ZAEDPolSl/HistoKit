# Example 1: Reading Whole Slide Images (WSIs)

---

In this example, we will demonstrate how to read **Whole Slide Images (WSIs)** using the **HistoKit** toolbox.

HistoKit provides efficient tools for handling large WSIs, allowing you to access and manipulate the image data effectively. It supports various formats, including:

- **SVS**
- **NDPI**
- **TIFF**
- **TIF**
- **JPEG**
- **PNG**
- **BMP**
- ... and many more.

Under the hood, HistoKit leverages **OpenSlide** and **PIL** libraries as backends to read and process these images efficiently.
